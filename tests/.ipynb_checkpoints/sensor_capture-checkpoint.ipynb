```python
import serial
import matplotlib.pyplot as plt

```

```python
# FSR/Flex capture

ser = serial.Serial('COM7', 115200, timeout=1)

data = []
i = 0
while i < 200:
    line = ser.readline().decode('utf-8').strip()
    
    if line:
        data.append(int(line))
        i += 1

plt.plot(data)

ser.close()
```

```python
# IMU Capture

CAPTURE_SECONDS = 10

ser = serial.Serial('COM7', 115200, timeout=1)

with open("imu.txt", "w") as file:
    data = []
    i = 0
    while i < CAPTURE_SECONDS * 20:
        line = ser.readline().decode('utf-8').strip()
        
        if line:            
            file.write(line + '\n')
            i += 1

file.close()
ser.close()
```

```python
# IMU Visualization

# Parse data
ax = []
ay = []
az = []
gx = []
gy = []
gz = []
with open("imu.txt", 'r') as file:
    for line in file:
        line = line.strip()
        vals = line.split(' ')

        ax.append(int(vals[0]))
        ay.append(int(vals[1]))
        az.append(int(vals[2]))
        gx.append(int(vals[3]))
        gy.append(int(vals[4]))
        gz.append(int(vals[5]))

file.close()

plt.figure(figsize=(500, 600))
plt.subplot(2, 3, 1)
plt.plot(ax)
plt.subplot(2, 3, 2)
plt.plot(ay)
plt.subplot(2, 3, 3)
plt.plot(az)
plt.subplot(2, 3, 4)
plt.plot(gx)
plt.subplot(2, 3, 5)
plt.plot(gy)
plt.subplot(2, 3, 6)
plt.plot(gz)
```
